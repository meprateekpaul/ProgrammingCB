### 2. Working with FASTA Files
### Reading FASTA Files
#### Single Sequence:

```python
from Bio import SeqIO

# Reading a single sequence from a FASTA file
record = SeqIO.read("example3.fasta", "fasta")
print(f"ID: {record.id}")
print(f"Sequence: {record.seq}")
```

```python
record
```

#### Multiple Sequences:

```python
from Bio import SeqIO

# Reading multiple sequences from a FASTA file
for record in SeqIO.parse("example.fasta", "fasta"):
    print(f"ID: {record.id}")
    print(f"Sequence: {record.seq}")
```

#### Writing FASTA Files

```python
from Bio.Seq import Seq
from Bio.SeqRecord import SeqRecord
from Bio import SeqIO

# Creating a sequence record
record = SeqRecord(Seq("ATGCGTACGTT"), id="Sample1", description="Example Sequence")

# Writing the sequence to a FASTA file
SeqIO.write(record, "output.fasta", "fasta")
```

#### Manipulating Sequences in FASTA Format
* Example: Reverse complement of a sequence:

```python
record.seq.reverse_complement()
```

```python
record.seq
```

```python
record.seq = record.seq.reverse_complement()
SeqIO.write(record, "output_revcomp.fasta", "fasta")
```

### 3. Working with GenBank Files
#### Reading GenBank Files
#### Single Sequence:

```python
from Bio import SeqIO

# Reading a single sequence from a GenBank file
record = SeqIO.read("sequence.gb", "genbank")
print(f"ID: {record.id}")
print(f"Sequence: {record.seq}")
```

```python
record.features
```

```python
record.description
```

### Multiple Sequences:

```python
from Bio import SeqIO

# Reading multiple sequences from a GenBank file
for record in SeqIO.parse("example4.gp", "genbank"):
    print(f"ID: {record.id}")
    print(f"Sequence: {record.seq}")
```

### Writing GenBank Files


```python
from Bio.Seq import Seq
from Bio.SeqRecord import SeqRecord
from Bio import SeqIO

# Creating a sequence record with the required molecule_type annotation
record = SeqRecord(Seq("ATGCGTACGTT"),
                   id="Sample1",
                   description="Example Sequence",
                   annotations={"molecule_type": "DNA"})

# Writing the sequence to a GenBank file
SeqIO.write(record, "output.gb", "genbank")
```

### Extracting Features from GenBank Files
#### Example: Extracting CDS features:

```python
for feature in record.features:
    if feature.type == "CDS":
        print(f"CDS Location: {feature.location}")
        print(f"CDS Sequence: {feature.extract(record.seq)}")
```

### 4. Advanced Sequence Operations
#### Translating DNA to Protein


```python
protein_seq = record.seq.translate()
print(f"Protein Sequence: {protein_seq}")
```

#### Calculating GC Content


```python
from Bio.SeqUtils import gc_fraction

gc_content = gc_fraction(record.seq) * 100  # Convert to percentage
print(f"GC Content: {gc_content}%")
```

### Converting Between Formats


```python
# Converting a GenBank file to FASTA format
SeqIO.convert("sequence.gb", "genbank", "example_converted.fasta", "fasta")
```

### 5. Q&A and Practice Examples:
##### Practice Problem 1:
Task: Read a GenBank file and extract all gene sequences, saving them in FASTA format.

```python
from Bio import SeqIO
from Bio.SeqRecord import SeqRecord

records = SeqIO.parse("sequence.gb", "genbank")
with open("genes.fasta", "w") as output:
    for record in records:
        for feature in record.features:
            if feature.type == "gene":
                gene_seq = feature.extract(record.seq)
                locus_tag = feature.qualifiers.get('locus_tag', ['unknown_locus'])[0]
                gene_record = SeqRecord(gene_seq, id=locus_tag, description="")
                SeqIO.write(gene_record, output, "fasta")
```

#### Practice Problem 2:
Task: Translate a DNA sequence from a FASTA file and write the protein sequence to a new file.

```python
record = SeqIO.read("output_revcomp.fasta", "fasta")
protein_seq = record.seq.translate()
protein_record = SeqRecord(protein_seq, id=record.id, description="Translated Protein")
SeqIO.write(protein_record, "protein_output.fasta", "fasta")
```
