##### Example 6: Calculating GC Content from a DNA Sequence File



````python
def calculate_gc_content(sequence):
    g = sequence.count('G')
    c = sequence.count('C')
    return (g + c) / len(sequence) * 100

with open('dna_sequences.txt', 'r') as file:
    flag=0
    for line in file:
        if flag==0:
          flag=1
          continue
        sequence = line.strip()

        gc_content = calculate_gc_content(sequence)
        print(f"Sequence: {sequence}, GC Content: {gc_content:.2f}%")
````

#### Example 8: Writing a FASTA File

````python
def write_fasta(sequences, output_file):
    with open(output_file, 'w') as file:
        for header, sequence in sequences.items():
            file.write(f">{header}\n")
            file.write(f"{sequence}\n")

sequences = {
    "seq1": "ATGCGTACGTT",
    "seq2": "CGTAGCTAGCTAG"
}

write_fasta(sequences, 'output.fasta')
````

### Practical: Building a Simple FASTA Parser

#### Example 9: Identifying Open Reading Frames (ORFs)



````python
def find_orfs(sequence):
    start_codon = 'ATG'
    stop_codons = ['TAA', 'TAG', 'TGA']
    orfs = []
    for i in range(len(sequence) - 2):
        if sequence[i:i+3] == start_codon:
            for j in range(i, len(sequence) - 2, 3):
                codon = sequence[j:j+3]
                if codon in stop_codons:
                    orfs.append(sequence[i:j+3])
                    break
    return orfs

fasta_sequences = read_fasta('dna_sequences.fasta')
for header, sequence in fasta_sequences.items():
    orfs = find_orfs(sequence)
    print(f">{header}\nORFs: {orfs}")
````

#### Example 10: Modifying Sequences and Saving to a New File

````python
def reverse_complement(sequence):
    complement = {'A': 'T', 'T': 'A', 'C': 'G', 'G': 'C'}
    return ''.join(complement[base] for base in reversed(sequence))

def modify_sequences(input_fasta, output_fasta):
    sequences = read_fasta(input_fasta)
    modified_sequences = {header: reverse_complement(seq) for header, seq in sequences.items()}
    write_fasta(modified_sequences, output_fasta)

modify_sequences('dna_sequences.fasta', 'modified_output.fasta')
````
