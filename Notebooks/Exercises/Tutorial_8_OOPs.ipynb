# Tutorials: Introduction to Object-Oriented Programming (OOP) in Python for Biologists

## 1. Introduction to Object-Oriented Programming (OOP)

### What is Object-Oriented Programming?

* OOP is a programming paradigm that uses "objects" to design applications and programs.
* Objects can represent real-world entities and can hold both data (attributes) and functions (methods).

### Importance of OOP in Python

* Python is a popular programming language in biology because of its simplicity and rich ecosystem of libraries.
* OOP allows for more organized and reusable code, which is particularly useful in managing complex biological data and models.

### Benefits of OOP for Biologists

* Encapsulation helps in managing biological data securely.
* Inheritance allows creating specialized biological models based on general ones.
* Polymorphism makes it easy to handle different biological entities in a uniform manner.


## 2. Basic Concepts of OOP
###  Classes and Objects

- A Class is a blueprint for creating objects (instances of a class).
- An Object is an instance of a class.
### Attributes and Methods

- Attributes: Variables that belong to the class.
- Methods: Functions that belong to the class.
### Instantiation of Objects

* The process of creating an object from a class.
#### Example: Modeling a DNA Sequence

````python
class DNA:
    def __init__(self, sequence):
        self.sequence = sequence

    def transcribe(self):
        return self.sequence.replace('T', 'U')

# Creating an object
dna_sequence = DNA("ATGCGTACGTT")
print(dna_sequence.sequence)  # Output: ATGCGTACGTT
print(dna_sequence.transcribe())  # Output: AUGCGUACGUU
````

## 3. Encapsulation and Abstraction
### Encapsulation: Protecting Data

* Encapsulation involves keeping the data (attributes) of a class private and only accessible through methods.
* This helps in protecting the integrity of the data.
### Abstraction: Simplifying Complex Reality

* Abstraction hides the complex reality and only shows the essential features.
* In biology, this can be used to model complex biological processes in a simplified manner.
### Example: Creating a Class for a Protein

````python
class Protein:
    def __init__(self, sequence):
        self.__sequence = sequence  # Private attribute

    def get_sequence(self):
        return self.__sequence

    def mutate(self, position, new_amino_acid):
        sequence_list = list(self.__sequence)
        sequence_list[position] = new_amino_acid
        self.__sequence = ''.join(sequence_list)

# Creating an object
my_protein = Protein("MVLSPADKTNVKAA")
print(my_protein.get_sequence())  # Output: MVLSPADKTNVKAA
my_protein.mutate(2, 'G')
print(my_protein.get_sequence())  # Output: MGLSPADKTNVKAA
````

## 4. Inheritance
### What is Inheritance?

* Inheritance allows a class to inherit attributes and methods from another class.
* The class that inherits is called a Derived Class, and the class being inherited from is called a Base Class.
### Base and Derived Classes

* A base class contains general attributes and methods.
* A derived class contains more specific attributes and methods.
### Example: Creating a Hierarchy of Organisms

````python
class Organism:
    def __init__(self, name):
        self.name = name

    def breathe(self):
        print(f"{self.name} is breathing.")

class Plant(Organism):
    def photosynthesize(self):
        print(f"{self.name} is photosynthesizing.")

class Animal(Organism):
    def move(self):
        print(f"{self.name} is moving.")

# Creating objects
fern = Plant("Fern")
fern.breathe()  # Output: Fern is breathing.
fern.photosynthesize()  # Output: Fern is photosynthesizing.

cat = Animal("Cat")
cat.breathe()  # Output: Cat is breathing.
cat.move()  # Output: Cat is moving.
````

## 5. Polymorphism
### What is Polymorphism?

* Polymorphism allows objects of different classes to be treated as objects of a common base class.
* It is particularly useful when different types of objects are expected to share the same behavior.
### Method Overriding

* Derived classes can override methods of the base class to provide specific behavior.
### Example: Modeling Different Types of Cells

````python
class Cell:
    def function(self):
        print("General cell function.")

class RedBloodCell(Cell):
    def function(self):
        print("Transporting oxygen.")

class Neuron(Cell):
    def function(self):
        print("Transmitting signals.")

# Using polymorphism
cells = [RedBloodCell(), Neuron(), Cell()]
for cell in cells:
    cell.function()

# Output:
# Transporting oxygen.
# Transmitting signals.
# General cell function.
````

## 6. Hands-on Practice: Building a Simple Biological Model
#### Building a Biological Class (e.g., Enzyme)

````python
class Enzyme:
    def __init__(self, name, function):
        self.name = name
        self.function = function

    def catalyze(self):
        print(f"{self.name} is catalyzing the reaction: {self.function}.")
````

#### Extending the Model with Inheritance (e.g., Specific Enzymes like Proteases)

````python
class Protease(Enzyme):
    def catalyze(self):
        print(f"{self.name}, a protease, is breaking down proteins.")
````

#### Using Polymorphism to Model Enzyme Reactions

````python
enzymes = [Protease("Trypsin", "Hydrolysis of peptide bonds"), Enzyme("Amylase", "Breakdown of starch")]

for enzyme in enzymes:
    enzyme.catalyze()

# Output:
# Trypsin, a protease, is breaking down proteins.
# Amylase is catalyzing the reaction: Breakdown of starch.
````
